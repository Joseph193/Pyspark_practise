```python
data = [ ('John', '2024-01-01', 100), ('John', '2024-01-05', 200), ('John', '2024-01-01', 400), ('John', '2024-01-05', 500),
         ('John', '2024-01-10', 300), ('Mary', '2024-01-01', 150), ('Mary', '2024-01-04', 250), ('Mary', '2024-01-09', 350) ]
schema ="Name string, Date string, amount int"
```

```python
df=spark.createDataFrame(data,schema)
df.display()
```

```python
Windowspec = Window.partitionBy("Name").orderBy(desc("Date"))
last_3_tran_df = df.withColumn("row_number", row_number().over(Windowspec)).filter(col("row_number") < 4)
last_3_tran_df.display()
```

```python
df=spark.createDataFrame(data,schema)
df.display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python
Windowspec=Window.partitionBy("Name").orderBy(desc("Date"))
lastest_trans_df=df.withColumn("Row_Number",row_number().over(Windowspec)).filter(col("Row_Number")<=3 )
lastest_trans_df.display()
```

```python
rolling_sum_df=lastest_trans_df.groupBy("Name").agg(sum("Amount").alias("Rolling_Sum"))
rolling_sum_df.display()
```

```python
Windowspec=Window.partitionBy("Name").orderBy("Date").rowsBetween(Window.unboundedPreceding, Window.currentRow)
rolling_sum_df=lastest_trans_df.withColumn("Rolling_Sum",sum("Amount").over(Windowspec))
rolling_sum_df.display()
```

```python
df = spark.sql('select * from sales_data')
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
```

```python
windowSpec=Window.partitionBy("region").orderBy("year").rowsBetween(Window.unboundedPreceding,Window.currentRow)
running_Total_df=df.withColumn("Running_Total",sum("Revenue").over(windowSpec))
running_Total_df.display()
```

```python
windowSpec=Window.partitionBy("region").orderBy("year").rowsBetween(Window.unboundedPreceding,Window.currentRow)
Lag_Total_df=df.withColumn("Lag_Percentage",lag("Revenue").over(windowSpec))
running_Total_df.display()
```

```python
#Year on Year Percentage Growth
df.display()
```

```python
sales_df=df.groupBy("region","year").agg(sum("revenue").alias("Total_Revenue"))
sales_df.display()
```

```python
windowSpec=Window.partitionBy("region").orderBy("year")
Previous=sales_df.withColumn("previous_Year",lag("Total_Revenue").over(windowSpec))
percent_df=Previous.withColumn("Percentage_Growth",(col("Total_Revenue")-col("previous_Year"))*100/col("previous_Year"))
percent_df.display()
```

```python
sales_df=df.groupBy("region","year").agg(sum("revenue").alias("Total_Revenue"))
sales_df.display()
```

```python
windowSpec=Window.partitionBy("region").orderBy("year").rowsBetween(-1,0)
Moving_df=sales_df.withColumn("2Row_Moving_Avg",avg("Total_Revenue").over(windowSpec))
Moving_df.display()
```
